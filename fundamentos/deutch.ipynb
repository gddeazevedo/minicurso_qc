```python
import ket
from ket.qulib.gates import unitary
```

```python
def phase_oracle(f, n):
    size = 2 ** n
    matrix = [[0 for _ in range(size)] for _ in range(size)]

    for i in range(size):
        matrix[i][i] = (-1) ** f(i)

    def inner(q):
        U = unitary(matrix)
        return U(q)

    return inner
```

# Algoritmo de Deutch

```python
def deutsch(oracle):
    proc = ket.Process()
    q = proc.alloc(1)

    ket.H(q)

    oracle(q)

    ket.H(q)

    result = ket.measure(q).get()
    proc.execute()

    if result == 0:
        print("The function is constant.")
    else:
        print("The function is balanced.")


def f_constant(x):
    return 0


def f_balanced(x):
    return 1 - x


deutsch(phase_oracle(f_constant, 1))
deutsch(phase_oracle(f_balanced, 1))
```

# Algoritmo de Deutch-Jozsa

```python
def deutsch_jozsa(oracle, n):
    proc = ket.Process()
    qubits = proc.alloc(n)
    ket.H(qubits)

    oracle(qubits)

    ket.H(qubits)

    result = ket.measure(qubits).get()

    if result == 0:
        print("The function is constant.")
    else:
        print("The function is balanced.")


def f_constant_n(x):
    return 0


def f_balanced_n(x):
    return bin(x).count("1") % 2


deutsch_jozsa(phase_oracle(f_constant_n, 1), 1)
deutsch_jozsa(phase_oracle(f_balanced_n, 1), 1)
```
